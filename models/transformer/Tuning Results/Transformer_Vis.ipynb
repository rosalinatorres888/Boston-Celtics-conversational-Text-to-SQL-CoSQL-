** Visualization of Model Hyparameters **

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("transformer_runs_data.csv").set_index("Run Number")

df.head(10)
```

```python
sns.set_theme(style='darkgrid')

fig, ax = plt.subplots(figsize=(6, 10))
sns.barplot(data=df, x="Run Number", y="BLEU Score", color="orange", label="Sub Value", width=0.4,)
sns.barplot(data=df, x="Run Number", y="Accuracy", color="royalblue", label="Accuracy", width=0.4, alpha=0.5)

# Label each set of bars
ax.bar_label(ax.containers[0],fmt="%.2f",color="darkorange",fontweight="bold")
ax.bar_label(ax.containers[1],fmt="%.2f",color="navy",fontweight="bold")

plt.ylim(0, 110)
plt.legend()
plt.show()
```
